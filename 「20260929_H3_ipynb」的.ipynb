```python
w = float(input("請輸入體重kg："))
h = float(input("請輸入身高cm："))
bmi=w/(h*0.01)**2
if bmi < 18.5:
    x= "過輕"
else:
  if bmi < 24:
    x= "正常"
  else:
    x= "過重"
print(f"你的 BMI 值為: {bmi:.2f}, 判定為{x}")
```
